```python
data_num = torchvision.datasets.MNIST(
    root='./data_num',
    train=True,
    download=True
)

data_num_small = Subset(data_num, range(10000))
```

```python
model = MLP(784, 10, 10, 10)
loss = Loss(10)
```

```python
optimizer = Optimizer(model.params(), lr = 0.001)
```

```python
image, label = data_num_small[10]
tensor([(p / 255.0) for p in image.getdata()])
```

```python
for epoch in range(150):
    for image, label in data_num_small:
        image = np.array([(p / 255.0) for p in image.getdata()])
        image = tensor(image)

        image = image.unsqueeze(0)
        print(image.shape)
        y_pred = model(image)
        y_train = [0]*10
        y_train[label] = 1
        y_train = tensor(y_train).unsqueeze(0)
        lo = loss.MSE(y_pred, y_train)
        optimizer.zero_grad()
        lo.backward()
        optimizer.step()
        print(lo)
    print(epoch)
```
